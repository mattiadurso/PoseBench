```python
time_data = {}
dataset = "imc"
for scene in os.listdir(f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/{mapper}/{dataset}/"):

    # COLMAP
    mapper = "colmap"
    time_path = f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/{mapper}/{dataset}/{scene}/time_taken.txt"
    with open(time_path, "r") as f:
        lines = f.readlines()
    data = {}
    for line in lines:
        line = line.strip()
        if ":" in line:
            key, value = line.split(":")
            key = key.strip()
            value = float(value.strip().replace("s", ""))
            key = f"{key}_{mapper}" if key == "Mapping" else key
            data[key] = value
    time_data[scene] = data

    # GLOMAP
    mapper = "glomap"
    time_path = f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/{mapper}/{dataset}/{scene}/time_taken.txt"
    with open(time_path, "r") as f:
        lines = f.readlines()
    
    for line in lines:
        line = line.strip()
        if ":" in line:
            key, value = line.split(":")
            key = key.strip()
            value = float(value.strip().replace("s", ""))
            key = f"{key}_{mapper}" if key == "Mapping" else key
            time_data[scene] = {**time_data[scene], key: value}

    # VGGT
    mapper = "vggt"
    time_path = f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/vggt/{dataset}/{scene}/sparse/timings.txt"
    try:
        with open(time_path, "r") as f:
            lines = f.readlines()

        for line in lines:
            line = line.strip()
            if ":" in line:
                key, value = line.split(":")
                key = key.strip()
                value = float(value.strip().replace("s", ""))
                key = f"{key}_vggt" if key == "total" else key
                time_data[scene] = {**time_data[scene], key: value}
    except FileNotFoundError:
        time_data[scene] = {**time_data[scene]}

    # MapAnything
    mapper = "map_anything"
    time_path = f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/map_anything/{dataset}/{scene}/timings.txt"
    try:
        with open(time_path, "r") as f:
            lines = f.readlines()

        for line in lines:
            line = line.strip()
            if ":" in line:
                key, value = line.split(":")
                key = key.strip()
                value = float(value.strip().replace("s", ""))
                key = f"{key}_map_anything" if key == "total" else key
                time_data[scene] = {**time_data[scene], key: value}
    except FileNotFoundError:
        time_data[scene] = {**time_data[scene]}
```

```python
import os
import pandas as pd
broken_images = {}

scenes = os.listdir("/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_2D/imc/data/phototourism/")
for scene in sorted(scenes):
    try:
        df_ = pd.read_csv(f"/home/mattia/Desktop/Repos/wrapper_factory/benchmarks_3D/results/vggt/imc_results_dfs/results_{scene}.csv")
    except Exception as e:
        print(f"Error processing {scene}: {e}")
        continue
    images = set(df_['image1'].tolist() + df_['image2'].tolist())
    broken_images_temp = []
    for image in images:
        # select tuplest with image1 or image2 equal to image image
        df_image = df_[(df_['image1'] == image) | (df_['image2'] == image)]
        # if all values in max error are inf
        if (df_image['max_error'] == float('inf')).all():
            broken_images_temp.append(image)
    broken_images[scene] = broken_images_temp
```

```python
import pycolmap
from joblib import Parallel, delayed
from typing import Dict, Set, Tuple, List
import numpy as np
from tqdm import tqdm

def extract_point3d_ids(image) -> Set[int]:
    """
    Extract all 3D point IDs visible in an image.
    
    Args:
        image: COLMAP image object
        
    Returns:
        Set of 3D point IDs
    """
    point3d_ids = set()
    for point2d in image.points2D:
        if point2d.has_point3D():
            point3d_ids.add(point2d.point3D_id)
    return point3d_ids


def precompute_image_point3d_sets(reconstruction: pycolmap.Reconstruction) -> Dict[int, Set[int]]:
    """
    Precompute 3D point sets for all images.
    
    Args:
        reconstruction: COLMAP reconstruction object
        
    Returns:
        Dictionary mapping image_id -> set of 3D point IDs
    """
    print("Precomputing 3D point sets for all images...")
    image_point3d_map = {}
    
    for image_id, image in tqdm(reconstruction.images.items(), desc="Extracting 3D points"):
        image_point3d_map[image_id] = extract_point3d_ids(image)
    
    # Debug: Print statistics
    total_points = sum(len(points) for points in image_point3d_map.values())
    avg_points = total_points / len(image_point3d_map) if image_point3d_map else 0
    print(f"Precomputed point sets for {len(image_point3d_map)} images")
    print(f"Total 3D points across all images: {total_points}")
    print(f"Average 3D points per image: {avg_points:.2f}")
    
    # Check if any images have points
    images_with_points = sum(1 for points in image_point3d_map.values() if len(points) > 0)
    print(f"Images with at least 1 3D point: {images_with_points}/{len(image_point3d_map)}")
    
    return image_point3d_map


def compute_connection_batch(
    image_pairs: List[Tuple[int, int]],
    image_point3d_map: Dict[int, Set[int]],
    min_shared_points: int = 50
) -> List[Tuple[int, int, int]]:
    """
    Compute connections for a batch of image pairs.
    
    Args:
        image_pairs: List of (image_id1, image_id2) tuples
        image_point3d_map: Precomputed mapping of image_id -> 3D point IDs
        min_shared_points: Minimum number of shared points to consider connected
        
    Returns:
        List of (image_id1, image_id2, num_shared_points) for connected pairs
    """
    connections = []
    
    for img_id1, img_id2 in image_pairs:
        points3d_img1 = image_point3d_map.get(img_id1, set())
        points3d_img2 = image_point3d_map.get(img_id2, set())
        
        num_shared = len(points3d_img1.intersection(points3d_img2))
        
        if num_shared >= min_shared_points:
            connections.append((img_id1, img_id2, num_shared))
    
    return connections


def find_connected_images_parallel(
    reconstruction: pycolmap.Reconstruction,
    min_shared_points: int = 50,
    n_jobs: int = -1,
    batch_size: int = 1000,
    debug: bool = False
) -> List[Tuple[int, int, int]]:
    """
    Find all connected image pairs using parallel processing.
    
    Args:
        reconstruction: COLMAP reconstruction object
        min_shared_points: Minimum number of shared 3D points to consider connected
        n_jobs: Number of parallel jobs (-1 uses all CPUs)
        batch_size: Number of image pairs to process per batch
        debug: Print debug information
        
    Returns:
        List of (image_id1, image_id2, num_shared_points) tuples
    """
    import time
    start_time = time.time()
    
    # Step 1: Precompute 3D point sets for all images
    image_point3d_map = precompute_image_point3d_sets(reconstruction)
    num_images = len(image_point3d_map)
    
    # Step 2: Generate all image pairs
    image_ids = sorted(image_point3d_map.keys())
    all_pairs = []
    for i in range(len(image_ids)):
        for j in range(i + 1, len(image_ids)):
            all_pairs.append((image_ids[i], image_ids[j]))
    
    total_pairs = len(all_pairs)
    print(f"\nTotal image pairs to check: {total_pairs}")
    
    # Debug: Check first few pairs manually
    if debug and total_pairs > 0:
        print("\n=== Debug: Checking first 10 pairs manually ===")
        for idx, (img_id1, img_id2) in enumerate(all_pairs[:10]):
            points1 = image_point3d_map[img_id1]
            points2 = image_point3d_map[img_id2]
            shared = len(points1.intersection(points2))
            print(f"Pair {idx+1}: Image {img_id1} ({len(points1)} pts) <-> "
                  f"Image {img_id2} ({len(points2)} pts) = {shared} shared")
    
    # Step 3: Split pairs into batches
    batches = [
        all_pairs[i:i + batch_size] 
        for i in range(0, len(all_pairs), batch_size)
    ]
    print(f"Split into {len(batches)} batches of ~{batch_size} pairs each")
    
    # Step 4: Process batches in parallel with progress bar
    print(f"Processing with {n_jobs} jobs...")
    results = Parallel(n_jobs=n_jobs, verbose=0)(
        delayed(compute_connection_batch)(
            batch, 
            image_point3d_map, 
            min_shared_points
        )
        for batch in tqdm(batches, desc="Processing batches")
    )
    
    # Step 5: Flatten results
    connections = []
    for batch_result in results:
        connections.extend(batch_result)
    
    elapsed_time = time.time() - start_time
    print(f"\nFound {len(connections)} connected image pairs (>= {min_shared_points} shared points)")
    print(f"Total time: {elapsed_time:.4f} s")
    
    # If no connections found, try with lower threshold
    if len(connections) == 0 and min_shared_points > 1:
        print(f"\n⚠️  No connections found with threshold {min_shared_points}")
        print("Checking with lower thresholds...")
        
        for test_threshold in [10, 5, 1]:
            test_connections = []
            for batch in batches:
                test_connections.extend(
                    compute_connection_batch(batch, image_point3d_map, test_threshold)
                )
            if test_connections:
                print(f"  ✓ Found {len(test_connections)} connections with threshold {test_threshold}")
                break
        else:
            print("  ✗ No connections found even with threshold 1")
    
    return connections


def build_connectivity_graph(
    connections: List[Tuple[int, int, int]]
) -> Dict[int, List[Tuple[int, int]]]:
    """
    Build an adjacency list representation of the connectivity graph.
    
    Args:
        connections: List of (image_id1, image_id2, num_shared_points) tuples
        
    Returns:
        Dictionary mapping image_id -> list of (connected_image_id, num_shared_points)
    """
    graph = {}
    
    for img_id1, img_id2, num_shared in connections:
        if img_id1 not in graph:
            graph[img_id1] = []
        if img_id2 not in graph:
            graph[img_id2] = []
        
        graph[img_id1].append((img_id2, num_shared))
        graph[img_id2].append((img_id1, num_shared))
    
    return graph


def print_connectivity_stats(
    connections: List[Tuple[int, int, int]],
    reconstruction: pycolmap.Reconstruction
):
    """Print statistics about image connectivity."""
    if not connections:
        print("\n⚠️  No connections found!")
        return
    
    num_shared_points = [c[2] for c in connections]
    
    print("\n=== Connectivity Statistics ===")
    print(f"Total images: {len(reconstruction.images)}")
    print(f"Connected pairs: {len(connections)}")
    print(f"Shared points - Min: {min(num_shared_points)}")
    print(f"Shared points - Max: {max(num_shared_points)}")
    print(f"Shared points - Mean: {np.mean(num_shared_points):.2f}")
    print(f"Shared points - Median: {np.median(num_shared_points):.2f}")
    
    # Build graph to find connectivity per image
    graph = build_connectivity_graph(connections)
    connections_per_image = [len(neighbors) for neighbors in graph.values()]
    
    if connections_per_image:
        print(f"\nConnections per image - Min: {min(connections_per_image)}")
        print(f"Connections per image - Max: {max(connections_per_image)}")
        print(f"Connections per image - Mean: {np.mean(connections_per_image):.2f}")
        print(f"Connections per image - Median: {np.median(connections_per_image):.2f}")


def analyze_reconstruction(reconstruction_path: str):
    """Analyze a reconstruction to understand why connections might be missing."""
    print(f"\n=== Analyzing Reconstruction ===")
    reconstruction = pycolmap.Reconstruction(reconstruction_path)
    
    print(f"Total images: {len(reconstruction.images)}")
    print(f"Total 3D points: {len(reconstruction.points3D)}")
    print(f"Total cameras: {len(reconstruction.cameras)}")
    
    # Check each 3D point's track length
    track_lengths = []
    for point3d_id, point3d in reconstruction.points3D.items():
        track_lengths.append(point3d.track.length())
    
    if track_lengths:
        print(f"\n3D Point Track Statistics:")
        print(f"  Min track length: {min(track_lengths)}")
        print(f"  Max track length: {max(track_lengths)}")
        print(f"  Mean track length: {np.mean(track_lengths):.2f}")
        print(f"  Median track length: {np.median(track_lengths):.2f}")
        
        # Points visible in multiple images
        multi_view_points = sum(1 for t in track_lengths if t >= 2)
        print(f"  Points visible in >=2 images: {multi_view_points}/{len(track_lengths)}")
    else:
        print("  ⚠️  No 3D points have tracks!")
    
    # Check images
    print(f"\nImage Statistics:")
    points_per_image = []
    for image_id, image in reconstruction.images.items():
        num_points = sum(1 for p2d in image.points2D if p2d.has_point3D())
        points_per_image.append(num_points)
    
    if points_per_image:
        print(f"  Min points per image: {min(points_per_image)}")
        print(f"  Max points per image: {max(points_per_image)}")
        print(f"  Mean points per image: {np.mean(points_per_image):.2f}")
        print(f"  Images with 0 points: {sum(1 for p in points_per_image if p == 0)}")
    
    return reconstruction
```

```python
# Configuration
reconstruction_path = "results/vggt/eth3d/electro/sparse"
min_shared_points = 50
n_jobs = -1  # Use all CPUs
batch_size = 1000
debug = True

# Analyze reconstruction first
reconstruction = analyze_reconstruction(reconstruction_path)

# Find connections
connections = find_connected_images_parallel(
    reconstruction,
    min_shared_points=min_shared_points,
    n_jobs=n_jobs,
    batch_size=batch_size,
    debug=debug
)

# Print statistics
print_connectivity_stats(connections, reconstruction)

# Build connectivity graph
if connections:
    graph = build_connectivity_graph(connections)
    
    # Display sample connectivity
    print("\n=== Sample Connectivity (first 5 images) ===")
    for img_id in sorted(graph.keys())[:5]:
        neighbors = graph[img_id]
        print(f"Image {img_id}: connected to {len(neighbors)} images")
        # Show top 3 strongest connections
        top_neighbors = sorted(neighbors, key=lambda x: x[1], reverse=True)[:3]
        for neighbor_id, num_shared in top_neighbors:
            print(f"  -> Image {neighbor_id}: {num_shared} shared points")
    
    # Save to JSON if needed
    import json
    output_data = {
        "connections": [
            {"image_id1": c[0], "image_id2": c[1], "shared_points": c[2]}
            for c in connections
        ],
        "min_shared_points": min_shared_points,
        "total_images": len(reconstruction.images)
    }
    
    output_path = "connections.json"
    with open(output_path, 'w') as f:
        json.dump(output_data, f, indent=2)
    print(f"\n✓ Connections saved to {output_path}")
```
